```python
# Convert ARFF to CSV using scipy
from scipy.io import arff
import pandas as pd

# Load ARFF file
data = arff.loadarff('/Users/adarsh/Phishing Websites Classification Project/phishing-classification/data/Training Dataset.arff')
df = pd.DataFrame(data[0])

# Convert bytes to strings if needed
for col in df.columns:
    if df[col].dtype == object:
        df[col] = df[col].str.decode('utf-8')

# Save as CSV
output_path = '/Users/adarsh/Phishing Websites Classification Project/phishing-classification/data/phishing_data.csv'
df.to_csv(output_path, index=False)

print(f"✓ CSV file created successfully!")
print(f"Shape: {df.shape}")
print(f"File saved: {output_path}")
print("\nFirst few rows of the dataset:")
print(df.head())
```
